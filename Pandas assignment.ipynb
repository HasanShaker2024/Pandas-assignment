# Explanation of data

## 1-total_bill:

Description: The total amount of the bill (in dollars) for the meal, including the cost of food and drinks.
Data Type: Float
Example: 16.99
## 2-tip:

Description: The amount of the tip (in dollars) given by the customer to the server.
Data Type: Float
Example: 1.01
## 3-sex:

Description: The gender of the person who paid for the meal.
Data Type: Categorical (string)
Possible Values: 'Male', 'Female'
Example: 'Female'
## 4-smoker:

Description: Indicates whether there were smokers in the party.
Data Type: Categorical (string)
Possible Values: 'Yes', 'No'
Example: 'No'
## 5-day:

Description: The day of the week when the meal was served.
Data Type: Categorical (string)
Possible Values: 'Thur' (Thursday), 'Fri' (Friday), 'Sat' (Saturday), 'Sun' (Sunday)
Example: 'Sun'
## 6-time:

Description: The time of day when the meal was served.
Data Type: Categorical (string)
Possible Values: 'Lunch', 'Dinner'
Example: 'Dinner'
## 7-size:

Description: The number of people in the party.
Data Type: Integer
Example: 2

## Question 10: What is the standard deviation of the total bill amounts?


````python
display(df["total_bill"].std()) # By default std dev is for sample
display(df.total_bill.std(ddof=1)) #For Sample
display(df.total_bill.std(ddof=0)) # For Population


````

## Question 11: How many male smokers are there compared to female smokers?


````python
display(df.groupby(["smoker","sex"]).agg({"smoker":"count"})) # Comparison for smokers and non smokers between male and femal


print("-"*30)
print("Using value.counts(): ")
display(df[df["smoker"]=="Yes"]["sex"].value_counts()) # only Smokers in male and femal


print("-"*30)
print("Another Way Using groupby() and size() ")
display(df.groupby(df[df["smoker"]=="Yes"]["sex"]).size())
````

## Question 12: How much more do male customers tip on average than female customers?

````python
#df.groupby("sex").agg({"tip":"mean"})

display(df.groupby("sex").agg({"tip":"mean"})) # Comparison between male and female average tip 

print("-"*30)
male_tip=df[df["sex"]=="Male"].agg({"tip":"mean"})
female_tip=df[df["sex"]=="Female"].agg({"tip":"mean"})
print("Average Male Tip - Average Female Tip :")
print(male_tip-female_tip)

print("-"*30)
print("Average Male Tip - Average Female Tip :")
display(df[df["sex"]=="Male"]["tip"].mean()-df[df["sex"]=="Female"]["tip"].mean()) # tip average difference

````

## Question 13: What is the median tip amount given by customers on Thursdays?

````python
display(df[df["day"]=="Thur"].agg({"tip":"median"}))

display(df[df["day"]=="Thur"]["tip"].median())
````

## Question 14: What is the total tip amount given by customers on Friday?


````python
display(df[df["day"]=="Fri"].agg({"tip":"sum"}))

display(df[df["day"]=="Fri"]["tip"].sum())
````

## Question 15: What is the average tip amount for non-smokers during dinner time?

````python
display(df[(df["smoker"]=="No")& (df["time"]=="Dinner")]["tip"].mean())


print("-"*30)
print("Another Way:")
# group data by "smoker" & "time" then calculate the mean of tip for each group 
# filtered to these values "No" (non smoker), and "Dinner" (time)
display(df.groupby(["smoker","time"])["tip"].mean().loc["No","Dinner"]) 
````
